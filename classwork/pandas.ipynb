```python
import pandas as pd

''' series has only one column in table but 
data frame has multiple columns per row '''

data={"Name" : ['rj','aarnavi', 'surabhi','lisha'],
      "age":   ['17','18','19','18'],
      "hobby": ['art','doomscroll','work','games']}

df=pd.DataFrame(data)
print(df)
print(df.columns)
print('------------------------------------------')
print(df.shape)
print('------------------------------------------')
print(df.head(2))
print(df.tail(2))
print('------------------------------------------')
print (df.info())
print('------------------------------------------')
print(df.describe())
print('------------------------------------------')
print(df.dtypes)
print(df['hobby'].value_counts())
```
