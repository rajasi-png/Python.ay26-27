```python
#assigment 5 string 
import re
def alphanumeric(S):
    return bool(re.fullmatch(r'[A-Za-z0-9@" "_-]+',S))

txt= input("enter a string")

if alphanumeric(txt):
    print("string passes")

else:
    print("string invalid")

```
